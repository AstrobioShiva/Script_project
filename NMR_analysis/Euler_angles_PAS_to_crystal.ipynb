```python
import numpy as np
from functions import Rabc
from sympy import *
```

```python
a1 = np.radians(153.5) #angles for crystal----> Goniometer Frame
b1 = np.radians(153.45)
g1 = np.radians(180)
R_crystal = Rabc(a1, b1, g1)

a = np.radians(308.4) #angles for Quad PAS -----> Goniometer frame
b = np.radians(32)
g = np.radians(65)

R_total = Rabc(a, b, g)
```

```python
R_PAS = np.matmul(np.linalg.inv(R_crystal), R_total)
print(R_PAS)

#calculated angles for Quad PAS -----> Crystal Frame
b_PAS = np.degrees(np.arccos(R_PAS[2,2]))
a_PAS = np.degrees(np.arctan(R_PAS[2,1]/R_PAS[2,0]))
g_PAS = np.degrees(np.arctan(-R_PAS[1,2]/R_PAS[0,2]))

print(a_PAS, b_PAS, g_PAS)
```

**Check if Eigen values remain same after rotation**

```python
A = np.zeros((3,3))

A[0,0] = 2; A[0,1] =-3; A[0,2] = 1;
A[1,0] = 1; A[1,1] = -2; A[1,2] = 1;
A[2,0] = 1; A[2,1] = -3; A[2,2] = 2;

print("Original matrix A:")
print(A, '\n')

eigenvalues, eigenvectors = np.linalg.eig(A)
print("Eigenvalues of A:")
print(eigenvalues.real, '\n')

print("Eigenvectors of A:")
print(eigenvectors.real, '\n')


#Rotation matrix
R_T = Rabc(np.radians(20), np.radians(30), np.radians(40))
print("Rotation matrix R:")
print(R_T, '\n')

#Rotated matrix = RAR'

A_rot = np.matmul(np.matmul(R_T, A),np.linalg.inv(R_T))
print("Rotated matrix A_rot:")
print(A_rot, '\n')

# Calculate and print the eigenvalues of the rotated matrix A_rot
eigenvalues_rot, eigenvectors_rot = np.linalg.eig(A_rot)
print("Eigenvalues of A_rot:")
print(eigenvalues_rot, '\n')

# print("Eigenvectors of A_rot:")
# print(eigenvectors_rot)

# A_rot = np.matmul()

```
