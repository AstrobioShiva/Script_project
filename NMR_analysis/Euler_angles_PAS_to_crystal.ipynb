```python
import numpy as np

from sympy import *
```

```python
def Rabc(alfa1, beta1, gama1): #Euler Rotation Matrix
    U = np.zeros((3, 3))
    #Changing input angles from degrees to radians
    alfa1 = np.radians(alfa1) 
    beta1 = np.radians(beta1)
    gama1 = np.radians(gama1)
    
    U[0, 0] = np.cos(alfa1)*np.cos(beta1)*np.cos(gama1) - np.sin(alfa1)*np.sin(gama1)
    U[0, 1] = np.sin(alfa1)*np.cos(beta1)*np.cos(gama1) + np.cos(alfa1)*np.sin(gama1)
    U[0, 2] = -np.sin(beta1)*np.cos(gama1)
    U[1, 0] = -np.cos(alfa1)*np.cos(beta1)*np.sin(gama1) - np.sin(alfa1)*np.cos(gama1)
    U[1, 1] = -np.sin(alfa1)*np.cos(beta1)*np.sin(gama1) + np.cos(alfa1)*np.cos(gama1)
    U[1, 2] = np.sin(beta1)*np.sin(gama1)
    U[2, 0] = np.cos(alfa1)*np.sin(beta1)
    U[2, 1] = np.sin(alfa1)*np.sin(beta1)
    U[2, 2] = np.cos(beta1)
    return U
```

```python
a1 = (153.5) #angles for crystal----> Goniometer Frame
b1 = (153.45)
g1 = (180)
R_crystal = Rabc(a1, b1, g1)

a = (308.4) #angles for Quad PAS -----> Goniometer frame
b = (32)
g = (65)

R_total = Rabc(a, b, g)
```

```python
R_PAS = np.matmul(np.linalg.inv(R_crystal), R_total)
print(R_PAS)

#calculated angles for Quad PAS -----> Crystal Frame
b_PAS = np.degrees(np.arccos(R_PAS[2,2]))
a_PAS = np.degrees(np.arctan(R_PAS[2,1]/R_PAS[2,0]))
g_PAS = np.degrees(np.arctan(-R_PAS[1,2]/R_PAS[0,2]))

print(a_PAS, b_PAS, g_PAS)
```

**L-Alanine**

```python
a = (333.5) # angles for crystal----> Goniometer Frame (from XRD)
b = (33.7)
g = (90.0)


R_crystal = Rabc(a, b, g)


angle_sets = {                              # angles for PAS(Q) -----> Goniometer frame (from ASICS)
    "Site1": [[48.0, 46.9, 336.1], 
               [309.9, 24.4, 102.4], 
               [128.7, 49.6, 203.3],
               [118.3, 6.9, 57.9]],
    
    "Site2": [[263.2, 9.1, 149.8], 
               [223.1, 1.0, 2.5], 
               [170.5, 38.5, 323.3],
               [145.3, 56.3, 37.5]]
}

for label, angles in angle_sets.items():
    print(f"\n===== Results for {label} =====\n")
    
    for i, ang in enumerate(angles):
        R_Lala_total = Rabc(ang[0], ang[1], ang[2])
        R_PAS_Lala = np.matmul(np.linalg.inv(R_crystal), R_Lala_total)

        # Use arctan2 to handle quadrant correctly
        z_PAS = np.degrees(np.arctan(R_PAS_Lala[2, 1]/ R_PAS_Lala[2, 0]))
        n_PAS = np.degrees(np.arctan(-R_PAS_Lala[1, 2] / R_PAS_Lala[0, 2]))
        l_PAS = np.degrees(np.arccos(R_PAS_Lala[2, 2]))

        print(f"Nucleus: {i+1},\n{R_PAS_Lala}\n")
        print(f"Euler angles for Quadrupolar tensor PAS → Tenon frame (L-alanine):")
        print(f" alpha_QG: {np.round(ang[0], 0)}, beta_QG: {np.round(ang[1], 0)}, gamma_QG: {np.round(ang[2], 0)}\n")
        print(f"Euler angles for Quadrupolar tensor PAS → Crystal frame (L-alanine):")
        print(f" zeta: {np.round(z_PAS, 0)}, lambda: {np.round(l_PAS, 0)}, nu: {np.round(n_PAS, 0)}\n")
```
