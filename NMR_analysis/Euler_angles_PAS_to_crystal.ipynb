```python
import numpy as np
from functions import Rabc
```

```python
a1 = np.radians(80) #angles for crystal----> Goniometer Frame
b1 = np.radians(45)
g1 = np.radians(14.5)
R_crystal = Rabc(a1, b1, g1)

a = np.radians(283.5) #angles for Quad PAS -----> Goniometer frame
b = np.radians(68)
g = np.radians(30)

R_total = Rabc(a, b, g)
```

```python
R_PAS = np.matmul(np.linalg.inv(R_crystal), R_total)
print(R_PAS)

#calculated angles for Quad PAS -----> Crystal Frame
b_PAS = np.degrees(np.arccos(R_PAS[2,2]))
a_PAS = np.degrees(np.arctan(R_PAS[2,1]/R_PAS[2,0]))
g_PAS = np.degrees(np.arctan(-R_PAS[1,2]/R_PAS[0,2]))

print(a_PAS, b_PAS, g_PAS)
```
