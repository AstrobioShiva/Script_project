**Define Rotation matrix**

*Reference: Interpretation of elements of the logarithm of a rotation matrix as rotation components around coordinate axes of a reference frame*

```python
import numpy as np
def Rabg(h1, k1, l1, phi1): #Euler Rotation Matrix for Euler angles Alpha(a), Beta (b), Gamma (g)
    U = np.zeros((3, 3))
    U[0, 0] = (1-h1**2)*np.cos(phi1) + h1**2
    U[0, 1] = h1*k1*(1-np.cos(phi1)) - l1*np.sin(phi1)
    U[0, 2] = l1*h1*(1-np.cos(phi1)) + k1*np.sin(phi1)
    U[1, 0] = h1*k1*(1-np.cos(phi1)) + l1*np.sin(phi1)
    U[1, 1] = (1-k1**2)*np.cos(phi1) + k1**2
    U[1, 2] = k1*l1*(1-np.cos(phi1)) - h1*np.sin(phi1)
    U[2, 0] = l1*h1*(1-np.cos(phi1)) - k1*np.sin(phi1)
    U[2, 1] = k1*l1*(1-np.cos(phi1)) - h1*np.sin(phi1)
    U[2, 2] = (1-l1**2)*np.cos(phi1) + l1**2
    return U
```

*Reference: Computing Euler angles from a rotation matrix (Gregory G. Slabaugh)
