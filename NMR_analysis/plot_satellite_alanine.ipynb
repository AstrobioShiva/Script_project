```python
path = "/home/shiva/WMU/PhD/Scripts/phd_project/Python/NMR/Data_exp/" 
file_start = 556                                       #enter the first file number 
file_end = 558                                         #enter the last file number
odd_file = [557]                                          #file with twice the scans

x_start_values = [238.6, -1467.7, 3836.5]                     # LEFT from data file
x_end_values = [-3449, -5155, 149]                            # RIGHT from data file

plt.figure(figsize=(15,5), dpi = 150)


# Load the data and normalize
for i, numb in enumerate(range(file_start, file_end + 1)):

    yData = np.loadtxt(path + "SW_Shiva" + str(numb) + ".txt", unpack=True)
    # print(f'file' "SW_Shiva" + str(numb), np.max(yData))
    if numb in odd_file:
        
        yData = yData*0.5           #taking half the scans (the acquisition time was double for this data collection) so plots can be merged        
        # print(np.max(yData))
    # yData = yData / np.max(yData)  # Normalize


    yData = yData - np.min(yData)  # Remove vertical offset
    # Calculate x-axis
    steps = len(yData)
    x_start = x_start_values[i]                              #value from data file
    x_end = x_end_values[i]                               #value from data file
    steps_ppm = (x_end - x_start)/steps
    x = np.arange(x_start, x_end, steps_ppm)
    
    plt.plot(x, yData, linewidth = 0.5, color = 'blue')
    
plt.xlim([4000, -5000])   

num_ticks = 15  # Adjust this for more or fewer ticks
xtick_positions = np.linspace(4000, -5000, num_ticks)  # Generate tick positions
plt.xticks(xtick_positions)  # Apply custom ticks 
plt.xlabel('Chemical Shift / ppm', fontsize = 16)
plt.ylabel('Amplitude', fontsize = 16)

# Save plot
plt.savefig('/home/shiva/WMU/PhD/Scripts/phd_project/Python/NMR/NMR_analysis/plots_experiment/L_alanine_satellite_xrot_30deg.png', dpi = 150 ,bbox_inches='tight')
plt.show()

```
