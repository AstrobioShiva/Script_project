**Find angle of triangle given two sides and one angle**

```python
import numpy as np
def find_angle(b,c, A):
    A = A*np.pi/180
    a = np.sqrt(b**2 + c**2 - 2*b*c*np.cos(A))
    return a
```

```python
b = find_angle(20.333, 25.1, 109.8360)
print(b)
```
