**L-alanine Results**

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd


plt.figure(figsize=(8,5), dpi=200)

df = pd.DataFrame({"id":["this study", "computed", "Pike et. al.", "Yamauchi et. al."], 
                 # Cq data
                   "Cq1Lala":[6.5, 8.26, 7.86, 8.1], 
                   "Cq2Lala":[6.6, 6.28, 6.53, 7.2],
                   
                   "sdCq1Lala":[0.4,0,0.05,0.3],
                   "sdCq2Lala":[0.4,0,0.05,0.3]}
                  )

plt.errorbar(np.arange(len(df['id'])), df['Cq1Lala'], yerr=df['sdCq1Lala'], ls = 'None', 
             marker='o', color = 'r', markersize=8, label = 'Site 1')
plt.errorbar(np.arange(len(df['id'])), df['Cq2Lala'], yerr=df['sdCq2Lala'], ls = 'None', 
             marker='o', color = 'b', markersize=8, label = 'Site 2')



ax = plt.gca()
ax.xaxis.set_ticks(np.arange(len(df['id'])))
ax.xaxis.set_ticklabels(df['id'], rotation=90, fontsize = 16)

# plt.xlabel("id", fontsize = 16)
plt.ylabel("Cq / MHz", fontsize = 16)

plt.legend(loc='upper left')
plt.show()
```
