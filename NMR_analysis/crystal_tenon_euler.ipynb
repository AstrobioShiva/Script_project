```python
import numpy as np
def get_euler_angles(a_or, b_or, c_or):
    b = np.degrees(np.arccos(c_or[2]))
    a = np.degrees(np.arctan(c_or[1]/c_or[0]))
    g = np.degrees(np.arctan(-b_or[2]/a_or[2]))

    return a, b, g

def get_direction_cosine(P, Q, R):
    

    crys_ten_a = [P[0]/np.sqrt(np.dot(P,P)), Q[0]/np.sqrt(np.dot(Q,Q)), R[0]/np.sqrt(np.dot(R,R))]

    crys_ten_b = [P[1]/np.sqrt(np.dot(P,P)), Q[1]/np.sqrt(np.dot(Q,Q)), R[1]/np.sqrt(np.dot(R,R))]

    crys_ten_c = [P[2]/np.sqrt(np.dot(P,P)), Q[2]/np.sqrt(np.dot(Q,Q)), R[2]/np.sqrt(np.dot(R,R))]

    return crys_ten_a, crys_ten_b, crys_ten_c


def rollett_matrix(a, b, c, al, be, ga):
    matrix = np.zeros((3,3))
    al = np.radians(al)
    be = np.radians(be)
    ga = np.radians(ga)
    
    cosga1 = (np.cos(al)*np.cos(be) - np.cos(ga))/(np.sin(al)*np.sin(be))
    singa1 = np.sqrt((1- cosga1**2))
    matrix[0,0] = a*np.sin(be)*singa1;
    matrix[1,0] = -a*np.sin(be)*cosga1; matrix[1,1] = b*np.sin(al)
    matrix[2,0] = a*np.cos(be); matrix[2,1] = b*np.cos(al); matrix[2,2] = c
    return matrix

def rollett_procedure (a,b,c):
    n = np.cross(b,c)
    b_orth = np.cross(c,n)
    a_orth = np.cross(b_orth, c)
    c_orth = c
    return a_orth, b_orth, c_orth
    

```
