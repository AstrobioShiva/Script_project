```python
import numpy as np

def dipolar(nucleus_i, nucleus_j, gamma_i, gamma_j, r_ij):
    """
    Function to calculate dipolar coupling between two nuclei
    
    Parameters:
    gamma_i (float): Nuclear spin of the first nucleus in rad*Hz/T
    gamma_j (float): Nuclear spin of the second nucleus in rad*Hz/T
    r_jk (float): Distance between the two nuclei in Angstroms

    Ref: 10.1002/cmr.a.20003 'Residual dipolar coupling constants: An elementary derivation of key equations'
    """
    h_bar = 1.05*10** -34                                       #Planck constant in J/Hz
    mu = 1.26*10** -6                                            #magnetic constant N A^-2
    k_ij =  -3*gamma_i*gamma_j*mu*h_bar*10**27 / (8*np.pi**2)   #k factor in kHz Angstrom^3
    dipolar_ij = (2/3)*(k_ij/r_ij**3)                            #dipolar coupling constant in kHz
    
    print(f"\nNuclei pair: {nucleus_i} - {nucleus_j}")
    print (f"k factor in kHzÅ³: {k_ij:.3f}" )
    print(f"Dipolar coupling in Hz: {dipolar_ij:.3f}")
```

```python
gamma_H = 26.68*10 **7
gamma_N = -2.712* 10**7
nucleus_H = "1H"
nucleus_N = "15N"
r_HH = 10
r_HN = 1.04

dipolar(nucleus_i=nucleus_H, nucleus_j= nucleus_H, gamma_i= gamma_H, gamma_j= gamma_H, r_ij = r_HH)

dipolar(nucleus_i=nucleus_H, nucleus_j= nucleus_N,gamma_i= gamma_H, gamma_j= gamma_N, r_ij = r_HN)

# print(k_HH, '\ndipolar coupling in Hz:', dipolar_HH)

# print(k_HN, '\ndipolar coupling in Hz:', dipolar_HN)

```
