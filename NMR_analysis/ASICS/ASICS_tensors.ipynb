```python
import numpy as np

#define CS Tensor function
def cs_tensor(delta_iso, delta, eta_cs):
    CS_PAS = np.zeros((3,3))
    CS_PAS[0,0] = delta_iso - (delta/2)*(1 + eta_cs)
    CS_PAS[1,1] = delta_iso - (delta/2)*(1 - eta_cs)
    CS_PAS[2,2] = delta_iso + delta

    return CS_PAS

#define Quadrupolar Tensor fucntion
def quad_tensor(CQ, eta_Q):
    Q_PAS = np.zeros((3,3))
    Q_PAS[0,0] = (-CQ/2)*(1-eta_Q)
    Q_PAS[1,1] = (-CQ/2)*(1+eta_Q)
    Q_PAS[2,2] = CQ

    return Q_PAS

```

```python
#Input values for ++8HQ (cluster1 , cluster2)
RHQ_delta_iso = [-8,-8]
RHQ_delta = [-9.08, -9.25]
RHQ_eta_cs = [0.88, 0.3]
RHQ_CQ = [2.18, 2.182]
RHQ_eta_Q = [0.935, 0.915]

for i, (delta_iso, delta, eta_cs, CQ, eta_Q) in enumerate(zip(RHQ_delta_iso, RHQ_delta, RHQ_eta_cs, RHQ_CQ, RHQ_eta_Q)):
    RHQ_CS_tensor  = cs_tensor(delta_iso, delta, eta_cs)
    RHQ_Q_tensor = quad_tensor(CQ, eta_Q)
    print(f'++8HQ CS Tensor Cluster{i+1}:\n', RHQ_CS_tensor)
    print(f'++8HQ Quadrupolar Tensor Cluster{i+1}:\n', RHQ_Q_tensor)
    print('==============================')

```

**Rotation Angles for Computed Tensor**

```python
#find eigenvalues and eigenvectors of original matrix
eigenvalues, eigenvectors = np.linalg.eig(HQ_CS_tensor) 

#CASTEP CS tensor (symmetric part)
HQ_CS_rot = np.zeros((3,3))                              #rotated CS tensor
HQ_CS_rot[0,0] = 90.0440; HQ_CS_rot[0,1] = -0.1909; HQ_CS_rot[0,2] = -1.44775;
HQ_CS_rot[1,0] = -0.1909; HQ_CS_rot[1,1] = 83.7563; HQ_CS_rot[1,2] = -4.55995;
HQ_CS_rot[2,0] = HQ_CS_rot[0,2]; HQ_CS_rot[2,1] = HQ_CS_rot[1,2]; HQ_CS_rot[2,2] = 78.0996
print('Rotated HQ_CS Tensor (Symmetric Part): \n', HQ_CS_rot)

eigenvalues_rot, eigenvectors_rot = np.linalg.eig(HQ_CS_rot)
print('\n Eigenvalues of rotated tensor: \n', eigenvalues_rot)
print('\n Eigenvectors of rotated tensor: \n', eigenvectors_rot)

R_T = np.matmul(eigenvectors_rot, np.linalg.inv(eigenvectors))
print("\n Calculated Rotation matrix:")
print(R_T,'\n')

b = np.degrees(np.arccos(R_T[2,2]))
a = np.degrees(np.arctan(R_T[2,1]/R_T[2,0]))
g = np.degrees(np.arctan(-R_T[1,2]/R_T[0,2]))
print("Calculated Euler angles (degrees):")
print(a, b, g,'\n')
```
