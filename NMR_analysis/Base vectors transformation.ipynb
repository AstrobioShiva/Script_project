```python
import numpy as np
#Matrix to transform base vectors a,b,c and beta(in degrees) to orthogonal set
#reference: Essentials of Crystallograohy McKie & McKie [Table 5.4]
def T(a,b,c, beta):
    U = np.zeros((3, 3))
    beta = beta*np.pi/180
    U[0,0] = a*np.sin(beta)
    U[1,1] = b
    U[2,0] = a*np.cos(beta)
    U[2,2] = c
    return U
```

```python
U = T(20.333,10.955,12.55,109.836) 
print(U)
```
