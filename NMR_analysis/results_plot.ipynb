```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Results for L-alanine
plt.figure(figsize=(10,8), dpi=300)

df = pd.DataFrame({"id":["this study", "computed", "exp1", "exp2"], 
                   "Cq1Lala":[6.5, 8.26, 7.86, 8.1], 
                   "Cq2Lala":[6.6, 6.28, 6.53, 7.2],
                   "sd1Lala":[0.4,0,0.05,0.3],
                   "sd2Lala":[0.4,0,0.05,0.3]}
                  )

plt.errorbar(np.arange(len(df['id'])), df['Cq1Lala'], yerr=df['sd1Lala'], ls = 'None', 
             marker='o', color = 'r', markersize=8, label = 'Site 1')
plt.errorbar(np.arange(len(df['id'])), df['Cq2Lala'], yerr=df['sd2Lala'], ls = 'None', 
             marker='o', color = 'b', markersize=8, label = 'Site 2')



ax = plt.gca()
ax.xaxis.set_ticks(np.arange(len(df['id'])))
ax.xaxis.set_ticklabels(df['id'], rotation=90, fontsize = 16)

# plt.xlabel("id", fontsize = 16)
plt.ylabel("Cq / MHz", fontsize = 16)

plt.legend(loc='upper left')
plt.show()
```
