```python
import itertools
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# Angles in degrees
angles = np.array([
    0, 10, 20, 30, 40, 50, 60, 70.1, 80, 90, 100, 109.9, 120, 130, 140, 150, 160, 170, 180,
    190, 200, 210, 220, 230, 240, 250.2, 260, 270, 280, 290, 300, 310, 320.1, 330, 340, 350, 360
])

# Original chemical shifts for Cluster 1
cluster1 = [
    [-26.7335, -28.2128, -29.5768], [-30.4335, -32.0881, -33.5802], [-29.3784, -31.0966, -32.5858],
    [-24.551, -25.995, -26.9512], [-17.8253, -19.2986, -22.4924], [-11.6432, -12.738, -17.2726],
    [-8.27888, -10.0123, -14.6161], [-8.35515, -10.962, -14.7313], [-10.3553, -13.6167, -16.7128],
    [-12.3765, -15.4362, -19.1978], [-13.3968, -17.0739, -20.161], [-11.9932, -16.8509, -19.9992],
    [-8.97628, -14.285, -17.6057], [-6.01214, -10.7574, -14.0134], [-4.84275, -8.28978, -11.534],
    [-7.19439, -9.43408, -12.0634], [-12.7489, -14.6673, -16.5417], [-19.8872, -21.5641, -23.0963],
    [-26.1043, -28.0453, -29.6712], [-29.9549, -31.7536, -33.4414], [-29.0379, -30.8431, -32.5114],
    [-24.2756, -26.3979, -28.0725], [-17.6195, -19.5632, -22.3637], [-11.9354, -13.7609, -17.2701],
    [-8.80409, -10.4365, -14.6902], [-8.31424, -10.6236, -15.092], [-10.5439, -13.9238, -17.5251],
    [-13.0558, -15.9649, -19.9101], [-14.4723, -17.629, -20.9454], [-12.9597, -16.1861, -19.3429],
    [-9.74227, -13.3151, -16.5399], [-6.53388, -10.0253, -13.3273], [-5.26836, -8.28776, -11.3684],
    [-7.63663, -9.61967, -12.4094], [-13.0454, -14.8504, -16.8331], [-20.1626, -21.6778, -23.2618],
    [-26.8229, -28.0662, -29.4727]
]

# Define the function to fit
def fit_func(theta, A, B, C):
    return A + B * np.cos(2 * np.radians(theta)) + C * np.sin(2 * np.radians(theta)) 

# Iterate over all permutations
permutations = list(itertools.permutations(range(3)))
best_fit = None
best_residuals = float('inf')
best_params = None
residual_cutoff = 5

for perm in permutations:
    # Apply permutation to cluster1 data
    perm_cluster1_0 = np.array([cluster[perm[0]] for cluster in cluster1])
    perm_cluster1_2 = np.array([cluster[perm[2]] for cluster in cluster1])
    
    # Calculate the differences
    diff_0_2 = perm_cluster1_0 - perm_cluster1_2
    
    # Fit the function to the data
    params, _ = curve_fit(fit_func, angles, diff_0_2)
    
    # Calculate the fitted values
    fitted_diff = fit_func(angles, *params)
    
    # Calculate residuals
    residuals = np.sum((diff_0_2 - fitted_diff) ** 2)
    print(residuals)
    
    # Check if this fit is better
    if residuals < 12:
        best_fit = (perm_cluster1_0, perm_cluster1_2)
        best_residuals = residuals
        best_params = params

        # Check if residuals are below the cutoff value
    if best_residuals < residual_cutoff:
        break


# Plot the best fit
plt.figure(figsize=(10, 6))
plt.plot(angles, best_fit[0] - best_fit[1], 'bo', label='Best Difference Cluster1_0 - Cluster1_2')
plt.plot(angles, fit_func(angles, *best_params), 'r-', label='Best Fitted Function')
plt.xlabel('Angle (degrees)')
plt.ylabel('Difference in Chemical Shift (ppm)')
plt.title('Best Fit for Difference between Cluster1_0 and Cluster1_2')
plt.legend()
plt.grid(True)
plt.show()

# Print the best fitting parameters and permutation
print(f'Best fitting parameters: A = {best_params[0]}, B = {best_params[1]}, C = {best_params[2]}' )
print(f'Best permutation: {permutations[np.argmin(best_residuals)]}')
```
