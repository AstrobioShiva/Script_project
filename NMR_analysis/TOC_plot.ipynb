```python
from libraries import *
path = "/home/shiva/WMU/PhD/Scripts/phd_project/Python/NMR/Data_exp/" 
yData1 = np.loadtxt(path + "SW_Shiva487.txt", unpack=True) #D-alanine spectrum
yData2 = np.loadtxt(path + "SW_Shiva402.txt", unpack=True) #L-alanine spectrum

yData1 = yData1 / max(yData1)  # Normalize
yData1 = yData1 - min(yData1)  # Remove vertical offset

yData2 = yData2 / max(yData2)  # Normalize
yData2 = yData2 - min(yData2)  # Remove vertical offset

# Calculate x-axis
steps1 = len(yData1)
x_start1 = 730.3                               #value from data file
x_end1 = -498.9                                #value from data file
steps_ppm1 = (x_end1 - x_start1)/steps1
x1 = np.arange(x_start1, x_end1, steps_ppm1)

steps2 = len(yData2)
x_start2 = 2451                               #value from data file
x_end2 = -1050.5                               #value from data file
steps_ppm2 = (x_end2 - x_start2)/steps2
x2 = np.arange(x_start2, x_end2, steps_ppm2)

plt.plot(x1, yData1, linewidth=0.9, color = 'magenta')
# Remove axes
plt.axis('off')
plt.xlim(300,30)

plt.savefig('/home/shiva/WMU/PhD/Scripts/phd_project/Python/NMR/NMR_analysis/plots_experiment/TOC_D_alanine.png', 
            dpi = 150,
            bbox_inches='tight', 
            transparent=True)
plt.show()

plt.plot(x2, yData2, linewidth = 0.9, color = '#008080')
# Remove axes
plt.axis('off')
plt.xlim(400,-50)
plt.savefig('/home/shiva/WMU/PhD/Scripts/phd_project/Python/NMR/NMR_analysis/plots_experiment/TOC_L_alanine.png', 
            dpi = 150,
            bbox_inches='tight', 
            transparent=True)
plt.show()
```
