```python
df = [0]*2
df[0] = pd.read_excel('/home/shiva/WMU/PhD/Scripts/phd_project/Python/NMR/Data_exp/Antisymmetric Chemical Shift Calculations.xlsx', sheet_name='Isotropic Change_X')
df[1] = pd.read_excel('/home/shiva/WMU/PhD/Scripts/phd_project/Python/NMR/Data_exp/Antisymmetric Chemical Shift Calculations.xlsx', sheet_name='Isotropic Change_Y')
```

```python
print(df[0])
print(df[1])
```

**Changing Electiric field in X direction**

```python
# Change of isotropic chemical shift (ppm) for Glycine with changing electric field
plt.figure(figsize=(10,8), dpi = 200)
plt.scatter(df[0].Electric_Field, df[0].Glycine, marker='o', s=80, color = 'b')
plt.xlabel('Electric Field / a.u.', fontsize=16)
plt.ylabel('Isotropic Chemical Shift / ppm', fontsize=16)

plt.show()
```

```python
# Change of isotropic chemical shift (ppm) for alanine with changing electric field
plt.figure(figsize=(10,8), dpi = 200)
plt.scatter(df[0].Electric_Field, df[0].R_Alanine, marker='o', s=80, color = 'g', facecolor = 'none', label = r'$\it{D-}$Alanine')
plt.scatter(df[0].Electric_Field, df[0].S_Alanine, marker='x', s=80, color = 'r', label = r'$\it{L-}$Alanine')
plt.xlabel('Electric Field / a.u.', fontsize=16)
plt.ylabel('Isotropic Chemical Shift / ppm', fontsize=16)
# Function add a legend

plt.legend()
plt.show()
```

**Changing Electiric field in Y direction**

```python
# Change of isotropic chemical shift (ppm) for Glycine with changing electric field
plt.figure(figsize=(10,8), dpi = 200)
plt.scatter(df[1].Electric_Field, df[1].Glycine, marker='o', s=80, color = 'b')
plt.title('Electric Field in X Direction', fontsize=18)
plt.xlabel('Electric Field / a.u.', fontsize=16)
plt.ylabel('Isotropic Chemical Shift / ppm', fontsize=16)
plt.show()
```
