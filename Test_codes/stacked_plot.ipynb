```python
import matplotlib.pyplot as plt
import numpy as np

# Example data
datasets = [
    {"x": np.linspace(0, 10, 100), "y": np.sin(np.linspace(0, 10, 100))},
    {"x": np.linspace(-5, 5, 150), "y": np.cos(np.linspace(-5, 5, 150))},
    {"x": np.linspace(1, 20, 200), "y": np.tan(np.linspace(1, 20, 200) * 0.1)},
]
labels = [r"$\sin(x)$", r"$\cos(x)$", r"$\tan(0.1x)$"]

# Create subplots
fig, ax = plt.subplots(figsize=(8, 12))  # Set sharey=True if y-axes need to align

# offset = 0
# Loop through datasets and plot
for i, (dataset, label) in enumerate(zip(datasets, labels)):
    offset= i*5
    #Normalize the data
    dataset["y"] = dataset["y"] / max(dataset["y"])

    ax.plot(dataset["x"], dataset["y"] + offset, label=label, linewidth=1.5)
    ax.legend(loc="upper right")
    ax.label_outer()
    # Increment the offset for the next dataset
    # offset += 2

# Add x-axis label only to the last subplot
ax.set_xlabel("Common X-Axis (e.g., ppm)")

ax.set_yticks([])  # Remove y-axis ticks
# Adjust spacing between subplots
plt.subplots_adjust(hspace=0.4)  # Increase/decrease for more/less space between subplots

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)
# Show the plot
plt.show()
```
